```python
import numpy as np
# The replace() function is used to convert "?" and " ?" into NaN so that Pandas can recognize them 
#as missing values and handle them during data cleaning.
df.replace("?", np.nan, inplace=True)
df.replace(" ?", np.nan, inplace=True)
```

```python
#drop = remove
#na = missing values (NaN)

#So dropna=True means "don't include missing values in the count." and false is oposite of it

#so That's why NaN will not shown.
```

```python
#df.dropna(inplace=True)
# used to remove row or columns,Used to remove missing values from the dataset.
#It performs data cleaning.
```

```python
print(df["workclass"].value_counts(dropna=False))
```

```python
# fill missing values
```

```python
print(df["workclass"].mode())
print(df["occupation"].mode())
print(df["native.country"].mode())
```

```python
df["workclass"]=df["workclass"].fillna(df["workclass"].mode()[0])
df["occupation"]=df["occupation"].fillna(df["occupation"].mode()[0])
df["native.country"]=df["native.country"].fillna(df["native.country"].mode()[0])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

1. Income Distribution (Bar Chart)

```python
income_counts = df["income"].value_counts()

plt.figure(figsize=(6,4))
plt.bar(income_counts.index, income_counts.values)

plt.title("Income Distribution")
plt.xlabel("Income")
plt.ylabel("Number of People")
plt.show()
```

2. Age Distribution (Histogram)

```python
plt.figure(figsize=(8,5))
plt.hist(df["age"], bins=20)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

3.Workclass Distribution (Bar Chart)

```python
workclass_counts = df["workclass"].value_counts()

plt.figure(figsize=(10,5))
plt.bar(workclass_counts.index, workclass_counts.values)

plt.title("Workclass Distribution")
plt.xlabel("Workclass")
plt.ylabel("Number of People")
plt.xticks(rotation=45)
plt.show()
```

4. Education Distribution (Bar Chart)

```python
education_counts = df["education"].value_counts()

plt.figure(figsize=(10,5))
plt.bar(education_counts.index, education_counts.values)

plt.title("Education Distribution")
plt.xlabel("Education")
plt.ylabel("Number of People")
plt.xticks(rotation=45)
plt.show()
```

5. Gender Distribution — Pie Chart

```python
gender_counts = df["sex"].value_counts()

plt.figure(figsize=(3,3))
plt.pie(gender_counts.values, labels=gender_counts.index, autopct="%1.1f%%")

plt.title("Gender Distribution")
plt.show()
```

6. Occupation Distribution — Bar Chart

```python
occupation_counts = df["occupation"].value_counts()

plt.figure(figsize=(5,5))
plt.bar(occupation_counts.index, occupation_counts.values)

plt.title("Occupation Distribution")
plt.xlabel("Occupation")
plt.ylabel("Number of People")
plt.xticks(rotation=90)
plt.show()
```

7. Hours per Week Distribution (Histogram)

```python
plt.figure(figsize=(8,5))

plt.hist(df["hours.per.week"], bins=20)

plt.title("Hours Worked Per Week")
plt.xlabel("Hours per Week")
plt.ylabel("Number of People")

plt.show()
```

8. Education vs Income (Count Plot)

```python
plt.figure(figsize=(10,6))

sns.countplot(data=df, x="education", hue="income")

plt.title("Education vs Income")
plt.xlabel("Education")
plt.ylabel("Count")
plt.xticks(rotation=90)
plt.show()
```

9. Correlation Heatmap

```python
plt.figure(figsize=(8,6))

sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

10: Age Box Plot

```python
plt.figure(figsize=(6,4))

plt.boxplot(df["age"])

plt.title("Age Box Plot")
plt.ylabel("Age")
plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.countplot(data=df, x="education", hue="income")

plt.title("Education Level vs Income")
plt.xlabel("Education")
plt.ylabel("Number of People")
plt.xticks(rotation=45)
plt.show()
```
